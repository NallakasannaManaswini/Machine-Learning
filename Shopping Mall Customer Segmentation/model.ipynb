PCA

```python
from sklearn.decomposition import PCA

pca=PCA(n_components=2)
X_df=pca.fit_transform(df)
```

```python
X_df
```

```python
X_df=pd.DataFrame(X_df)
```

```python
X_df.head()
```

KMeans

```python
from sklearn.cluster import KMeans
wcss=[]

for k in range(2,10):
    kmeans=KMeans(n_clusters=k,random_state=0,n_init=12)
    kmeans.fit(X_df)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

Elbow Method

```python
plt.plot(range(2,10),wcss,marker='o')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

kmean=KMeans(n_clusters=4,random_state=0,n_init=12)
new_col=kmean.fit_predict(X_df)
```

```python
new_col=pd.DataFrame(new_col,columns=['Cluster'])
```

```python
acc_score=silhouette_score(X_df,new_col)
```

```python
plt.scatter(X_df.iloc[:,0],X_df.iloc[:,1],c=new_col.values)
plt.show()
```

```python
import warnings
warnings.filterwarnings('ignore')
```

# The Silhouette Score is a metric used to evaluate the quality of clustering.The score ranges from -1 to 1, where:

### 1 indicates well-clustered points (clear separation).

### 0 suggests overlapping clusters.

### -1 means poor clustering (points are likely in the wrong cluster).

```python

for k in range(2, 15):
    kmeans = KMeans(n_clusters=k, random_state=0, n_init=12)
    new_columns = kmeans.fit_predict(X_df)
    score = silhouette_score(X_df, new_columns)
    print(f'K={k}, Silhouette Score={score}')
```
