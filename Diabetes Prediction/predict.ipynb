```python
import pickle
import numpy as np
import pandas as pd
```

```python
oe_model=pickle.load(open('oe.pkl','rb'))
logistic_model=pickle.load(open('logistic.pkl','rb'))
```

```python
data=[42,1,27.1,109,56,5.06,4.73,1.02,1.15,2.82,11.5,2.98,80.2,5.2,'Occational','Occational']
```

```python
data
```

```python
data_14=oe_model.transform([[data[14]]]).flatten().tolist()
data_15=oe_model.transform([[data[15]]]).flatten().tolist()
```

```python
data=data[:14]+data_14+data_15
```

```python
X_input=np.array(data,dtype=np.float32).reshape(1,-1)
```

```python
prediction = logistic_model.predict(X_input)[0]
```

```python
print("Prediction:","Diabetic" if prediction==1 else "Non-Diabetic")
```
